```python
t = int(input())

for _ in range(t):
    A, B, C, D, E = map(int, input().split())
    
    total = (
        A * 35034 +
        B * 23090 +
        C * 19055 +
        D * 12530 +
        E * 18090
    )
    
    # 센트를 달러로 변환
    dollars = total / 100
    
    print(f"${dollars:.2f}")
```
